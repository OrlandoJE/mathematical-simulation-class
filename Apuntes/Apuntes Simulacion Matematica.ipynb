```python
import numpy as np
import sympy as sym
import matplotlib.pyplot as plt
from ipywidgets import *
import pandas as pd
```

## Semana 2 Miércoles

```python
x = np.linspace(-1,1,100)

y1 = np.sin(x)
y2 = np.cos(x)
y3 = 1 - np.exp(x)
y4 = x**2
```

```python
plt.plot(x,y1, 'r--',label='sin')
plt.plot(x,y2, 'k:',label='cos')
plt.plot(x,y3,label='exp')
plt.plot(x,y4,label='xˆ2')
plt.legend(loc=4, fontsize=14)
plt.grid()
```

```python
fig, ax = plt.subplots(2,2)
ax[0,0].plot(x,y1)
ax[0,0].grid()
ax[1,0].plot(x,y2)
ax[1,0].grid()
ax[0,1].plot(x,y3)
ax[0,1].grid()
ax[1,1].plot(x,y4)
ax[1,1].grid()
```

```python
def grafica(rangoX, f):
  x = np.linspace(*rangoX, 100)
  y = f(x)
  plt.plot(x, y)
  plt.grid()
  plt.show()
```

```python
grafica((-np.pi,np.pi), np.sin)
```

```python
interact(grafica,rangoX=widgets.FloatRangeSlider(value=(-np.pi,np.pi),min=-10,max=10,step=0.01),f=[np.sin, np.cos])
```

```python
def g(x):
  return (x-2)**4
x = sym.var('x')
dg = sym.diff(g(x), x)
ddg = sym.diff(dg,x)
c = sym.solve(dg)
ddg.subs(x,c[0])
```

```python
x_num = np.linspace (1,3,500)
plt.plot(x_num,g(x_num))
plt.plot(c[0], g(c[0]), '*r', ms=10)
plt.grid()
```

```python
def h(x):
  return ((x-3)**2)*((x+3)**2)
x = sym.var('x')
dh = sym.diff(h(x), x)
ddh = sym.diff(dh,x)
c = sym.solve(dh)
ddh.subs(x,c[0])
x = np.array([1,c[2],4])
print(h(x))

x_num = np.linspace (1,4,500)
plt.plot(x_num,h(x_num))
plt.plot(c[2], g(c[2]), '*r', ms=10)
plt.grid()
```
