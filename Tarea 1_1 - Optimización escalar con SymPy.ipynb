# Ejercicios - Optimización escalar con SymPy


```python
import numpy as np
import sympy as sym
import matplotlib.pyplot as plt
```

Para la siguientes funciones encuentre los puntos críticos, grafique y diga si éste se trata de un máximo o mínimo, local o global.


1. $f(x)=x+\frac{1}{x^2}$


```python
def f(x):
    return x + 1/(x**2)
x = sym.var('x', real=True)
solutions = sym.solve(sym.diff(f(x), x))
x_num = np.linspace (min(solutions) - 1, max(solutions) + 1, 500)
plt.plot(x_num,f(x_num))
for solution in solutions:
    if (sym.diff(sym.diff(f(x), x), x).subs(x, solution) > 0):
        print(f"Minimo global en x = {solution}")
    elif (sym.diff(sym.diff(f(x), x), x).subs(x, solution) < 0):
        print(f"Máximo global en x = {solution}")
    else:
        print("Indeterminado")
    plt.plot(solution, f(solution), '*r', ms=10)
    
plt.grid()
```

4. $f(x)=\frac{15x}{2}\sqrt{\frac{x}{x-30}}$


```python
def f(x):
    return (15*x/2) * sym.sqrt(x /(x - 30))
x = sym.Symbol('x', real=True)
solutions = [solution for solution in sym.solve(sym.diff(f(x), x)) if solution > 0]
x_num = np.linspace (min(solutions) - 1, max(solutions) + 1, 500)
plt.plot(x_num, np.vectorize(lambda x: f(x).evalf())(x_num))
for solution in solutions:
    if (sym.diff(sym.diff(f(x), x), x).subs(x, solution) > 0):
        print(f"Minimo global en x = {solution}")
    elif (sym.diff(sym.diff(f(x), x), x).subs(x, solution) < 0):
        print(f"Máximo global en x = {solution}")
    else:
        print("Indeterminado")
    plt.plot(solution, f(solution), '*r', ms=10)
    
plt.grid()
```

5. $f(x)=6-2x+\sqrt{18-2x^2}$, para $-3\leq x\leq 3$

6. $f(x)=\frac{1}{4}x^4+\frac{1}{3}x^3-x^2$ en el intervalo $[-3,1.7]$


```python
def f(x):
    return x**4/4 + x**3/3 - x**2
x = sym.Symbol('x', real=True)
solutions = [solution for solution in sym.solve(sym.diff(f(x), x)) if solution >= -3 and solution <= 1.7]
x_num = np.linspace (min(solutions) - 1, max(solutions) + 1, 500)
solutions
plt.plot(x_num, np.vectorize(lambda x: f(x).evalf())(x_num))
for solution in solutions:
    if solution == -2:
        if (sym.diff(sym.diff(f(x), x), x).subs(x, solution) > 0):
            print(f"Minimo gobal en x = {solution}")
        plt.plot(solution, f(solution), '*r', ms=10)
    else:
        if (sym.diff(sym.diff(f(x), x), x).subs(x, solution) > 0):
            print(f"Minimo local en x = {solution}")
        elif (sym.diff(sym.diff(f(x), x), x).subs(x, solution) < 0):
            print(f"Máximo local en x = {solution}")
        else:
            print("Indeterminado")
        plt.plot(solution, f(solution), '*r', ms=10)
    
plt.grid()
```
